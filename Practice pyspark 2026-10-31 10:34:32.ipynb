```python
data = [
    (101, "Roshan", "Pune", "Maharashtra", 60000, "2026-10-25 10:00:00"),
    (102, "Rahul", "Mumbai", "Maharashtra", 45000, "2026-10-26 11:00:00"),
    (103, "Amit", "Delhi", "Delhi", 70000, "2026-10-27 09:00:00"),
    (101, "Roshan", "Pune", "Maharashtra", 65000, "2026-10-28 12:00:00"),
    (104, "Priya", "Nashik", "Maharashtra", 55000, "2026-10-29 14:00:00"),
    (102, "Rahul", "Mumbai", "Maharashtra", 48000, "2026-10-30 15:00:00"),
    (105, "Sneha", "Bangalore", "Karnataka", 80000, "2026-10-30 16:00:00")
]
columns=[
    "id",
    "name",
    "city",
    "state",
    "salary",
    "update_at"
]
customer_df = spark.createDataFrame(data,columns)
customer_df.show()
```

```python
from pyspark.sql.functions import col,when
#Salary >= 60000 High
#Salary < 60000	Low
customer_df.withColumn(
    "salary_category",
    when
        (col("salary")>=60000,"high")
    .otherwise("low")
).show()



```

```python
#>= 70,000	High
#>= 50,000 and < 70,000	Medium
#< 50,000

df =customer_df.withColumn(
    "salary_category",
    when
    (col("salary")>=70000,"high")
.when(
    (col("salary")>=50000)&(col("salary")<70000),
    "medium"
    )
.otherwise("low")

)
df.show()
```

```python
from pyspark.sql.functions import col ,lit,coalesce
df = customer_df.withColumn(
    "salary",
    coalesce(col("salary"),lit(0))

)
df.show()
```

Replace NULL values:
- salary → 0
- city → "Unknown"
- state → "Unknown"

```python
df= customer_df.withColumn(
    "salary",
    coalesce(col("salary"),lit(0))
).withColumn(
    "city",
    coalesce(col("city"),lit("unknown").alias("new_city"))
).withColumn(
    "state",

    coalesce(col("state"),lit("unknown").alias("new_city"))
)
df.show()

df = customer_df.select(
    "*",
    coalesce(col("salary"),lit(0)),
    coalesce(col("city"),lit("unknown").alias("new_city")),
    coalesce(col("state"),lit("unknown").alias("new_state"))

)
df.show()
```

```python
df = customer_df.select(
    "*",
    coalesce(col("salary"),lit(0)).alias("new_salary"),
    coalesce(col("city"),lit("unknown")).alias("new_city"),
    coalesce(col("state"),lit("unknown")).alias("new_state")

)
df.show()
```
