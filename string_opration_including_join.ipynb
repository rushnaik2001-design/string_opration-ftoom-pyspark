## # start string operation of pyspark on databrics

```python
data = [
    (101, "Roshan", "Pune", "Maharashtra", 60000, "2026-10-25 10:00:00"),
    (102, "Rahul", "Mumbai", "Maharashtra", 45000, "2026-10-26 11:00:00"),
    (103, "Amit", "Delhi", "Delhi", 70000, "2026-10-27 09:00:00"),
    (101, "Roshan", "Pune", "Maharashtra", 65000, "2026-10-28 12:00:00"),
    (104, "Priya", "Nashik", "Maharashtra", 55000, "2026-10-29 14:00:00"),
    (102, "Rahul", "Mumbai", "Maharashtra", 48000, "2026-10-30 15:00:00"),
    (105, "Sneha", "Bangalore", "Karnataka", 80000, "2026-10-30 16:00:00")
]

columns = [
    "customer_id",
    "customer_name",
    "city",
    "state",
    "salary",
    "updated_at"
]

customer_df = spark.createDataFrame(data, columns)

from pyspark.sql.functions import to_timestamp, col

customer_df = customer_df.withColumn(
    "updated_at",
    to_timestamp(col("updated_at"))
)

customer_df.printSchema()
customer_df.show(truncate=False)
```

Create a new column called name_length that contains the number of characters in each customer's name.

```python
from pyspark.sql.functions import col,length
df=customer_df.withColumn(
    "name_length",
    length(col("customer_name"))
)
df.show()
```

Your company wants to maintain the original customer name and create a separate uppercase version.

Task:
1. Create a new column called upper_name.
2. Convert name into uppercase.
3. Keep the original name column unchanged.
4. Display id, name, and upper_name.

```python
from pyspark.sql.functions import upper,lower,col
df = customer_df.withColumn(
    "upper_name",
    upper(col("customer_name"))

)
df = df.select(
    "customer_id",
    "customer_name",
    "upper_name",
)

df.show()
```

Your company wants to calculate the length of every customer's name.

Task:
1. Create a new column called name_length.
2. Calculate the number of characters in customer_name.
3. Keep the original columns unchanged.
4. Display customer_id, customer_name, and name_length.

```python
df=customer_df.withColumn(
    "name_length",
    length(col("customer_name"))
)
df=df.select(
    "customer_id","customer_name","name_length"
)
df.show()
```

Using the existing customer_df:

Remove leading and trailing spaces from the customer_name column.

```python
from pyspark.sql.functions import trim
df = customer_df.withColumn(
    "customer_name",
    trim(col("customer_name"))
)
df.select(
    "customer_id",
    "customer_name"
).show()
```
